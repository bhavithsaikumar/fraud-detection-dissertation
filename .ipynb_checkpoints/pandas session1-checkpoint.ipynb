```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "amount": [200, 1500, 800, 9000, 50],
    "location": ["London", "Lagos", "London", "Lagos", "London"],
    "fraud": [0, 1, 0, 1, 0]
}

df = pd.DataFrame(data)
print(df)
```

```python
df["fraud"].value_counts().plot(kind="bar")
plt.title("Fraud vs Non-Fraud")
plt.show()
```

```python
df["fraud"].value_counts().plot(kind="bar")
plt.title("Fraud vs Non-Fraud")
plt.xlabel("Fraud (0 = No, 1 = Yes)")
plt.ylabel("Count")
plt.show()
```

```python
df["amount"].plot(kind="hist", bins=5)
plt.title("Distribution of Transaction Amounts")
plt.xlabel("Amount")
plt.ylabel("Frequency")
plt.show()
```

```python
import seaborn as sns

# Only number columns can be correlated
sns.heatmap(df[["amount", "fraud"]].corr(), annot=True, cmap="coolwarm")
plt.title("Feature Correlation")
plt.show()
```

```python
df["fraud"].value_counts().plot(kind="bar")
plt.title("Fraud vs Non-Fraud")
plt.xlabel("Fraud")
plt.ylabel("Count")
plt.savefig("fraud_chart.png", dpi=300, bbox_inches="tight")
plt.show()
```
